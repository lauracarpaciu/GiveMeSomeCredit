A confusion matrix is a table that is often used to describe the performance of a classification model (or "classifier") on a set of test data for which the true values are known.

TN = 34983 : We predicted no, and they don't have the financial distress

FN = 2442 : We predicted no, but they actually do have the financial distress. (Also known as a "Type II error.")

FP = 34: We predicted yes, but they don't actually have the financial distress. (Also known as a "Type I error.")

TP =41 : These are cases in which we predicted yes (they have the financial distress), and they do have the financial distress.

Accuracy: Overall, how often is the classifier correct?
(TP+TN)/total

Precision: When it predicts yes, how often is it correct?
TP/predicted yes

True Positive Rate: When it's actually yes, how often does it predict yes?

TP/actual yes 
also known as "Sensitivity" or "Recall"

f1 score = 2*Recall*Precision/ Recall + Precision
